### Fonction `range()`
Exécuter les codes suivants. Que se passe-t-il ? 

````python
# on définit trois entiers, a, b, c
a, b, c = 3, 9, 2
print("a =", a)
print("b =", b)
print("c =", c)
````

````python
for i in range(a):
    print("i =", i)
````

````python
for i in range(a, b):
    print("i =", i)
````

````python
for i in range(a, b, c):
    print("i =", i)
````

### Itéreration sur une séquence : plusieurs méthodes
Exécuter les codes suivants. Expliquer ce qu'il se passe dans chaque cas.

#### Itération sur les éléments d'une séquence

````python
fruits = ["pomme", "kiwi", "cassis"]  # que renvoie type(fruits) ?
n_fruits = len(fruits)  # que contient n_fruits ?

for fruit in fruits:
    print(fruit)
````

````python
vowels = ["a", "e", "i", "o", "u"]
word = "Halloween"
for letter in word:
    if letter in vowels:
        print(letter)  # qu'affiche-t-on ?
````

#### Itération sur les indices d'une séquence
En quoi les codes A et B sont-ils différents ?

````python
# code A
word = "Python"
length = len(word)
for i in range(length):
    print(word[i])
````

````python
# code B
word = "Python"
for letter in word:
    print(letter)
````

#### Itération sur les indices d'une séquence : cas d'une liste

````python
animals = ["panda", "owl", "bear", "sheep", "dog"]
n_animals = len(animals)
for i in range(n_animals):
    print(animals[i])
````

### Utiliser un compteur

Exécuter et expliquer les codes ci-dessous. En quoi une boucle `for` diffère d'une boucle `while` ?

````python
# version while
word = "Pumpkin"
length = len(word)

i = 0
sum_p = 0

while i < length:

    if word[i] == "p":
        sum_p += 1

    i += 1

print(f"The word {word.lower()} has {sum_p} p.")
````

````python
# version for
word = "Pumpkin"
length = len(word)

sum_p = 0

for i in range(length):
    if word[i] == "p":
        sum_p += 1

print(f"The word {word.lower()} has {sum_p} p.")
````
